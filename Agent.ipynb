```python
class LinkSummary(BaseModel):
    title: str = Field(description="The title of the source")
    url: str = Field(description="The source URL")
    summary: str = Field(description="A concise summary of the source")


class ResearchResponse(BaseModel):
    topic: str = Field(description="The topic the user asked to research")
    links: list[LinkSummary] = Field(description="Relevant links with summaries")
```
